```python
# Import the Pandas library
# Pandas is used for creating and manipulating DataFrames

import pandas as pd
```

Create Employee Dataset


```python
# Create a dictionary containing employee information

data={
    "Emp_ID":[101,102,103,104,105,106,107,108],
    "Name":["Rahul","Priya","Amit","Riya","Karan","Neha","Vivek","Pooja"],
    "Department":["IT","HR","IT","Finance","HR","Finance","IT","HR"],
    "City":["Surat","Navsari","Surat","Valsad","Surat","Navsari","Valsad","Surat"],
    "Salary":[45000,38000,52000,48000,40000,51000,60000,42000],
    "Experience":[2,1,5,4,2,6,8,3]
}
```

```python
# Convert the dictionary into a Pandas DataFrame

df = pd.DataFrame(data)

df
```

Step 1:Display the dataset

```python
# Display all rows and columns of the DataFrame

df
```

Step 2:Filter Data

Display Employees from IT Department

```python
# Display employees who belong to the IT department

df[df["Department"]=="IT"]
```

Display Employees with Salary Greater Than 50,000

```python
# Display employees whose salary is greater than 50,000

df[df["Salary"]>50000]
```

Display Employees From Surat

```python
# Display employees who are working in Surat

df[df["City"]=="Surat"]
```

Apply Multiple Values

```python
# Apply multiple conditions:
# Display IT employees whose salary is greater than 50,000

df[(df["Department"]=="IT") & (df["Salary"]>50000)]
```

Step 3: Perform Sorting


Sort by Salary

```python
# Sort employees according to their salary
# By default, sorting is done in ascending order

df.sort_values(by="Salary")
```

Sort by ascending order

```python
# Sort employees according to salary in descending order
# Highest salary will appear first

df.sort_values(by="Salary",ascending=False)
```

Sort by Department and Salary

```python
# Sort data first by Department
# Then sort employees within each department by Salary
# Salary is sorted in descending order

df.sort_values(by=["Department","Salary"],ascending=[True,False])
```

Step 4:Perform Aggregation

Calculate Total Salary

```python
# Calculate the total salary of all employees

df["Salary"].sum()
```

Calculate Average Salary

```python
# Calculate the average salary of all employees

df["Salary"].mean()
```

Find highest Salary

```python
# Find the highest salary in the dataset

df["Salary"].max()
```

Find Lowest Salary

```python
# Find the lowest salary in the dataset

df["Salary"].min()
```

Count Employees

```python
# Count the total number of employees

df["Emp_ID"].count()
```

Step 5:Perform GroupBy Operations

Average Salary Department-Wise

```python
# Calculate the average salary for each department

df.groupby("Department")["Salary"].mean()
```

Minimum Salary Department-Wise

```python
# Find the minimum salary in each department

df.groupby("Department")["Salary"].min()
```

Count the Employees Department-Wise

```python
# Count the number of employees in each department

df.groupby("Department")["Emp_ID"].count()
```

Total Salary Department-wise

```python
# Calculate the total salary for each department

df.groupby("Department")["Salary"].sum()
```

Maximum Salary Department-wise

```python
# Find the maximum salary in each department

df.groupby("Department")["Salary"].max()
```

Minimumn Salary Department-wise

```python
# Find the manimum salary in each department

df.groupby("Department")["Salary"].min()
```

Group by City

```python
# Group employees according to their city
# Calculate the average salary for each city

df.groupby("City")["Salary"].mean()
```

Step 6:Multiple Aggregation Functions

```python
# Perform multiple calculations on salary department-wise
# count = number of employees
# sum   = total salary
# mean  = average salary
# max   = highest salary
# min   = lowest salary

df.groupby("Department")["Salary"].agg(["count","sum","mean","max","min"])
```

Step 7:Save CSV file

```python
# Save the employee DataFrame into a CSV file
# index=False prevents Pandas from adding an extra index column

df.to_csv("Employee_Data.csv",index=False)

# Display a message after successfully creating the CSV file

print("CSV File Created successfully.")
```
