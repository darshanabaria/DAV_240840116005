Practical-2: Perform data inspection, cleaning, handling missing values, duplicate removal, and data transformation using Pandas.


Step 1:Import the Pandas Library

```python
import pandas as pd
```

Step 2:Create Sample Data

```python
# Create a dictionary containing sample student data

data={
    "ID": [101,102,103,104,104,105,106,107],
    "Name": ["Rahul","Karan","Riya",None,"Amit","Priya","Krisha","Harry"],
    "Age": [20,21,None,19,20,22,24,28],
    "City": ["Surat","Navsari","Valsad","Bharuch","Surat","Navsari","Valsad","Surat"],
    "Marks": [85,None,78,88,88,87,90,93]
}
```

Step 3:Create a DataFrame

```python
# Convert dictionary into a DataFrame

df = pd.DataFrame(data)

# Display the original dataset

df
```

Step 4: Check Missing Values


Task 8:Check missing values in each columns.

```python
# Count missing values in each column

df.isnull().sum()
```

Task 9:Replace missing value in the marks columns.

```python
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

df
```

Task 10:Replace missing value in the age columns.

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())

df
```

Task 11:Replace missing names with the text "Unknown".

```python
df["Name"] = df["Name"].fillna("Unknown")

df
```

Step 5: Check Duplicate Records


Task 12:Check whether each row is duplicated.

```python
df.duplicated()
```

Task 13:Count the total number of duplicate rows.

```python
# Count duplicate rows

df.duplicated().sum()
```

Step 6: Remove Duplicate Records


Task 14:Remove the duplicate records.

```python
# Remove duplicate rows permanently

df.drop_duplicates()

df
```

Step 7: Data Cleaning


Task 15:Rename column "Marks" to "Score"

```python
# Rename Marks column to Score

df.rename(columns={"Marks":"Score"},inplace=True)

df
```

Step 8: Data Transformation


Task 16:Convert the age columns from float to integer.

```python
# Convert Age column into integer datatype

df["Age"] = df["Age"].astype(int)

df.dtypes
```

Task 17:Create a new column named "Final Score".

```python
# Increase every student's score by 5 marks

df["Final Score"] = df["Score"] + 5

df
```

Task 18:Convert All students names to uppercase.

```python
# Convert Name to uppercase

df["Name"] = df["Name"].str.upper()

df
```

Task 19:Convert All students city to lowercase.

```python
# Convert City names to lowercase

df["City"] = df["City"].str.lower()

df
```

Task 20:Remove leading and trailing spaces from city names.

```python
# Remove extra spaces from Name column

df["City"] = df["City"].str.strip()

df
```

Task 21:Sort the dataframe based on the "Final Score" column.

```python
df.sort_values(by="Final Score",ascending=False)
```

Step 9: Save Cleaned Dataset


Task 22:Save the cleaned DataFrame as a new CSV File.

```python
# Save cleaned dataset into a new CSV file

df.to_csv("cleaned_students.csv",index=False)

print("Dataset saved successfully")
```
