**Step 1: Import Libraries and Create Dataset**

```python
# Import pandas library.
import pandas as pd

# Import NumPy Library.
import numpy as np

# Import Matplot Library.
import matplotlib.pyplot as plt
```

```python
# Create Dataset.
data = {
    "Month" : ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"],
    "Product" : ["Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop","Mobile", "Tablet"],
    "Region" : [ "North", "South", "East", "West", "North", "South", "East", "West", "North", "South", "East", "West"],
    "Sales": [ 120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000, 250000],
    "Units_Sold" : [12, 30, 25, 14, 35, 30, 18, 38, 32, 21, 45, 40]
}
df=pd.DataFrame(data)
df
```

**Step 3: Create a Line Chart**

```python
# How did the company sales change from january to december?
plt.figure(figsize=(10, 5))
plt.plot(df["Month"], df["Sales"], marker="o", color="blue")
plt.title("Monthly Sales Trend (Jan to Dec)")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

**Step 3: Create a Bar Chart**


```python
# Which Product generates how much total sales.
product_sales = df.groupby("Product")["Sales"].sum()

plt.figure(figsize=(7, 5))
plt.bar(product_sales.index, product_sales.values, color=["brown", "orange", "pink"])
plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.show()
```

**Step 4: Create a Histogram**

```python
# How are the indivisual sales values distributed?
plt.figure(figsize=(7, 5))
plt.hist(df["Sales"], bins=5, color="pink", edgecolor="black")
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

**Step 5: Create a scatter plot**


```python
# Is there a relationship between the no. of unit_sold and total sale?
plt.figure(figsize=(7, 5))
plt.scatter(df["Units_Sold"], df["Sales"], color="brown")
plt.title("Units Sold vs Sales")
plt.xlabel("Units Sold")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

**Step 6: Create a Pie chart**


```python
# What percentage of total sales comes from each digit?
region_sales = df.groupby("Region")["Sales"].sum()

plt.figure(figsize=(6, 6))
plt.pie(region_sales.values, labels=region_sales.index, autopct="%1.1f%%", startangle=90)
plt.title("Sales Share by Region")
plt.show()
```
